```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
epochs = [1, 2, 3, 4, 5]
training_loss = [3.403664, 3.107256, 3.046429, 3.010378, 2.983080]
validation_loss = [3.142253, 3.077725, 3.051202, 3.040141, 3.034861]
accuracy = [0.212542, 0.223560, 0.227491, 0.228902, 0.232216]
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

ax1.plot(epochs, training_loss, label='Training Loss', color='green', linestyle='dashed')
ax1.plot(epochs, validation_loss, label='Validation Loss', color='green')
ax1.set_title('Training & Validation Loss')
ax1.set_xlabel('Epoch')
ax1.set_ylabel('Loss')
ax1.legend()

ax2.plot(epochs, accuracy, label='Accuracy', color='steelblue')
ax2.set_title('Accuracy')
ax2.set_xlabel('Epoch')
ax2.set_ylabel('Accuracy')
ax2.legend()

plt.tight_layout()
plt.show()
```
